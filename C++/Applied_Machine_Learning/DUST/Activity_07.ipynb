```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# 1. Create the given dataset as a Pandas DataFrame

data = {
    "Employee": ["E1", "E2", "E3", "E4", "E5", "E6"],
    "Age": [24, 30, 27, 35, 29, 32],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance"],
    "Experience": [2, 6, 4, 10, 5, 8],
    "Training Hours": [15, 25, 20, 30, 18, 28],
    "Projects Completed": [4, 8, 6, 12, 7, 10]
}

df = pd.DataFrame(data)


# 2. Display first five rows, shape, column names and data types

print("First Five Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)


# 3. Display the statistics on the data

print("\nStatistics:")
print(df.describe())


# 4. Scale the Age column using Min-Max scaling
#    and store the result in Age Scaled

scaler = MinMaxScaler()

df["Age Scaled"] = scaler.fit_transform(df[["Age"]])


# 5. Standardize the Experience column using StandardScaler
#    and store the result in Experience Standardized

scaler = StandardScaler()

df["Experience Standardized"] = scaler.fit_transform(
    df[["Experience"]]
)


# 6. Encode the Gender column using label encoding
#    Female = 0, Male = 1

df["Gender Encoded"] = df["Gender"].map({
    "Female": 0,
    "Male": 1
})


# 7. Encode the Department column using one-hot encoding

df = pd.get_dummies(
    df,
    columns=["Department"],
    dtype=int
)


# 8. Create Productivity Score
#    Training Hours + (2 x Projects Completed)

df["Productivity Score"] = (
    df["Training Hours"] +
    (2 * df["Projects Completed"])
)


# 9. Create Experience Level
#    0 to 3  = Beginner
#    4 to 7  = Intermediate
#    8 or more = Experienced

def experience_level(years):
    if years <= 3:
        return "Beginner"
    elif years <= 7:
        return "Intermediate"
    else:
        return "Experienced"


df["Experience Level"] = df["Experience"].apply(
    experience_level
)


# 10. Display the completely transformed dataset

print("\nCompletely Transformed Dataset:")
print(df)
```
