```python
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
advertising = np.array([1,2,3,4,5]) 
sales = np.array([35, 40, 50, 55, 65]) 
```

### Reshaping Dataset for LinearRegression Model

```python
X = advertising.reshape(-1,1)
y = sales
```

```python
plt.scatter(advertising, sales, label="Actual Data")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Visualizationof dataset")
plt.legend()
plt.show()
```

## Object creation

```python
model = LinearRegression()
```

#### Training a Linear Regression Model

```python
model.fit(X,y)
```

#### Coefficient and Intercept

```python
print("Coefficient: ",model.coef_[0])
print("Intercept: ",model.intercept_)
```

#### Regression Lines plotting

```python
y_pred = model.predict(X)
```

```python
plt.scatter(X, y_pred, label="Processed Data")
plt.plot(X, y_pred, label="Regression_Line")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Regression Line")
plt.legend()
plt.show()
```

```python
mae = mean_absolute_error(y,y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)
```

```python
print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("r2 score: ",r2)
```

```python
x_new = np.array([[55]])
prediction = model.predict(x_new)
print(prediction)
```

```python
x2_new = np.array([25,45,75]).reshape(-1,1)
predicition = model.predict(x2_new)
print(predicition)
```

```python
#For every 1-unit increase in advertising expenditure, 
#predicted sales increase by approximately 0.67 units, on average.
coefficient = model.coef_[0]
print("Coefficient: ", coefficient)
print(
    f"For every 1 unit increase in advertising expenditure, "
    f"sales are expect to increase approximately {coefficient:.2f} units"
)
print("Intercept: ",model.intercept_)
```
