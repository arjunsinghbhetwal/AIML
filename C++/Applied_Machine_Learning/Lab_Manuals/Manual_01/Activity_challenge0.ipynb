### Challenge Activity


```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
advertising = np.array([10, 20, 30, 40, 50, 60, 70, 80])
sales = np.array([25, 30, 38, 45, 50, 58, 65, 72])
```

```python
X = advertising.reshape(-1,1)
Y = sales

print(X)
print(Y)
```

### Visualization

```python
plt.scatter(X, Y, label="Actual Data")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Data Visualization")
plt.show()
```

#### Model Training

```python
model = LinearRegression()
model.fit(X,Y)
```

#### Coefficient and Intercept

```python
print("Coeff: ",model.coef_[0])
print("Intercept: ",model.intercept_)
```

```python
y_pred = model.predict(X)
print(y_pred)
```

#### Regression Line Plot

```python
plt.scatter(X, Y, label="Predicted Data")
plt.plot(X, y_pred, label="Regression Line")
plt.xlabel("Advertising")
plt.ylabel("Predicted_Sales")
plt.title("Regression Plot")
plt.legend()
plt.show()
```

####  Performance Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score 
```

```python
mae = mean_absolute_error(Y,y_pred)
mse = mean_squared_error(Y,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y,y_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2 score: ",r2)
```

### Sales Prediction 55

```python
X1 = np.array([[55]])
y1_pred = model.predict(X1)
print("Sales Prediciton for X1: ",y1_pred[0])
```

#### Sales Prediciton 25, 45 and 75

```python
X2 = np.array([25,45,75]).reshape(-1,1)
y2_pred = model.predict(X2)
print("Sales Prediction for X2: ",y2_pred)
```

#### Interpret Coefficient

```python
print("Interpretation")
print("For every 1 unit increase in advertising expenditure, "
      "the predicted sales increase approximately",
      model.coef_[0])
```
