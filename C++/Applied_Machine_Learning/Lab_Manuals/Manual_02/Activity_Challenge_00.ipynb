### Challenge Activity

```python
import numpy as np
```

```python
advertising = np.array([10, 20, 30, 40, 50, 60, 70, 80])
sales = np.array([25, 30, 38, 45, 50, 58, 65, 72])
```

```python
X = advertising.reshape(-1,1)
Y = sales

print(X.shape)
print(Y.shape)
```

### Visualization

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(X, Y, label="Actual Data")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Original Data")
plt.show()
```

### Model Creation

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(X,Y)
```

### Coefficient and Intercept

```python
print("Coefficient:",model.coef_[0])
print("Intercept:",model.intercept_)
```

### Plot Regression Line

```python
y_pred = model.predict(X)
print(y_pred)
```

```python
plt.scatter(X, Y, label="Actual Data")
plt.plot(X,y_pred,label = "Regression Line")
plt.xlabel("Advertising")
plt.ylabel("Sales")
plt.title("Regression Plot")
plt.legend()
plt.show()
```

### Performance Metrics

```python
from sklearn.metrics import mean_squared_error, mean_squared_error, r2_score
```

```python
mae = mean_squared_error(Y,y_pred)
mse = mean_squared_error(Y,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y,y_pred)
```

```python
print("MAE:",mae)
print("MSE:",mse)
print("RMSE:",rmse)
print("R2 Score:",r2)
```

### Sales Prediciton

```python
new_value = np.array([55]).reshape(-1,1)
y1_pred = model.predict(new_value)
print(y1_pred)
```

```python
new_values = np.array([25,45,75]).reshape(-1,1)
y2_pred = model.predict(new_values)
print(y2_pred)
```

```python
print("Interpretation")
print("For every 1 unit increase in advertising expenditure, "
      "the predicted sales increase approximately",
      model.coef_[0])
```
