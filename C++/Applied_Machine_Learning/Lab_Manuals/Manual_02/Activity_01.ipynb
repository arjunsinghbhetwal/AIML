### Activity 1

```python
import numpy as np
```

```python
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([35, 40, 50, 55, 65, 70, 75, 85])
```

```python
print("Hours:", hours)
print("Marks:", marks)
```

```python
print("Mean Hours: ",np.mean(hours))
print("Mean Marks: ",np.mean(marks))
print("Standard Deviation: ",np.std(marks))
print("Minimum Marks: ",np.min(marks))
print("Maximum Marks: ",np.max(marks))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(hours, marks)
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours Studied vs Marks")
plt.show()
```

```python
m, c = np.polyfit(hours,marks,1)
print("Slope:",m)
print("Intercept:",c)
```

```python
predicted_marks = m * hours + c
print("Predicted Marks:",predicted_marks)
print(predicted_marks)
```

```python
new_hours = 6.5
predicted = m * new_hours + c
print("Predicted Marks: ",predicted)
```

#### Task:Modifying program for 4 diff values

```python
hour_01 = np.array([2.5,4.5,7.5,10])
pred = m * hour_01 + c
print("Predicted Marks: ",pred)
```

### Visualization

```python
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted_marks, label="Regression Line")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Regression Line")
plt.legend()
plt.show()
```

```python
error = marks - predicted_marks
mse = np.mean((marks - predicted_marks)**2)
rmse = np.sqrt(mse)

print("Error: ",error)
print("MSE: ",mse)
print("RMSE: ",rmse)
```
