```python
df = pd.DataFrame({ 
"Marks": [40, 50, None, 70, 80], 
"City": ["Delhi", "Delhi", "Mumbai", None, "Delhi"] 
})

print(df)
```

#### Fill the missing values

```python
df["Marks_mean"] = df["Marks"].fillna(df["Marks"].mean())
df["Marks_median"] = df["Marks"].fillna(df["Marks"].median())
df["Marks_mode"] = df["Marks"].fillna(df["Marks"].mode()[0])

print(df)
```

#### Fill missing Values

```python
df = pd.DataFrame({
    "Temperature": [30, None, 34, None, 38]
})
```

```python
df["Forward_Fill"] = df["Temperature"].ffill()
df["Backward_Fill"] = df["Temperature"].bfill()
df["Interpolated"] = df["Temperature"].interpolate()

print(df)
```

### Min Max Scaling

```python
df = pd.DataFrame({"Age":[20,30,40,50]})
```

```python
df["Age_MinMax"] = (df["Age"] - df["Age"].min())/(df["Age"].max() - df["Age"].min())
print(df)
```

### Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.DataFrame({"Marks":[50,60,70,80]})
```

```python
scaler = StandardScaler()
```

```python
df["Marks_Standardized"] = scaler.fit_transform(df[["Marks"]])
print(df)
```

### One-Hot encoding

```python
df = pd.DataFrame({"City":["Delhi","Mumbai","Chennai","Delhi"]})
```

### Orinal Encoding

```python
df = pd.DataFrame({"Education":["School","Graduate","Postgraduate","Graduate"]})
```

```python
order = {"School":1, "Graduate":2, "Postgraduate":3}
df["Education_encoded"] = df["Education"].map(order)
```

```python
print(df)
```

### Feature Engineering

```python
df = pd.DataFrame({ 
"Maths": [70, 60, 90], 
"Science": [80, 65, 85], 
"English": [75, 70, 95] 
}) 
```

```python
df["Total"] = df["Maths"]+df["Science"]+df["English"]
df["Average"] = df["Total"]/3
```

```python
print(df)
```

### Extract Date Features

```python
df = pd.DataFrame({"Purchase_Date": ["2026-01-15", "2026-02-20", "2026-03-05"]}) 
```

```python
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"])
```

```python
df["Year"] = df["Purchase_Date"].dt.year
df["Month"] = df["Purchase_Date"].dt.month
df["Day"] = df["Purchase_Date"].dt.day
```

```python
print(df)
```

### 9. Experienced Level

```python
def experience_level(years):
    if years <= 3:
        return "Begineer"
    elif 4 <= years <= 7:
        return "Intermediate"
    else:
        return "Experienced"
    
df["Experienced_Level"] = df["Experience"].apply(experience_level)
print(df)
```
