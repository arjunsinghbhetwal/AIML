```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
x = np.linspace(0, 5, 11)
y = x ** 2
```

```python
help(plt)
```

### Basic Matplot lib commands


```python
plt.plot(x, y, 'g')
plt.title('String Title here')
plt.xlabel('X Axis Title')
plt.ylabel('Y Axis Title')
plt.show()
```

### Creating Multiplots on the same line

```python
plt.subplot(1,2,1)
plt.plot(x, y, 'r--')
plt.subplot(1,2,2)
plt.plot(y, x, 'g*-')
```

### Matplotlib Object Oriented Method

```python
#Create Figue(empty canvas)
fig = plt.figure()

axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) #left, bottom, width, height (range 0 to 1)

#Plot on that set of axes
axes.plot(x, y, 'b')
axes.set_xlabel('Set X label')
axes.set_ylabel('Set Y label')
axes.set_title('Set Title')
```

### Plot inside Plot

```python
#Create blank canvas
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8]) #main axes
axes2 = fig.add_axes([0.2, 0.5, 0.4, 0.3]) #inset axes

#Large figure Axes1 Figure 
axes1.plot(x, y, 'b')
axes1.set_xlabel('X label')
axes1.set_ylabel('Y label')
axes1.set_title('Axes 1 Title')

#Insert Figure Axes2
axes2.plot(x, y, 'r')
axes2.set_xlabel('X label')
axes2.set_ylabel('Y label')
axes2.set_title('Axes 2 Title')
```

### Subplots as Objects


```python
fig, axes = plt.subplots()

axes.plot(x, y, 'r')
axes.set_xlabel('X label')
axes.set_ylabel('Y label')
axes.set_title('Title_1')
```

```python
#specifying the no. of rows and columns when creating subplots() objects.
#Empty canvas of 1 by 2 subplots

fig, axes = plt.subplots(nrows=1, ncols=2)
```

```python
axes
```

### Iterating through the array:

```python
for ax in axes:
    ax.plot(x, y, 'b')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_title('Title')

fig #displays the figure object
```

### TO PREVENT THE OVERALAPPING OF THE PLOTS


```python
fig, axes = plt.subplots(nrows=1, ncols=2)

for ax in axes:
    ax.plot(x, y, 'r')
    ax.set_xlabel('X Label')
    ax.set_ylabel('Y Label')
    ax.set_title('title')

fig
plt.tight_layout() #auto adjusts the overlapping of the plots.
```
