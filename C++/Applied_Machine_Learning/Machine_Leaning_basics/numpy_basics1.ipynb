```python
import numpy as np

prices = np.array([29.99, 49.99, 19.99])
quantities = np.array([1,2,3])
```

```python
print(prices)
print(quantities)
```

```python
#Calculate Revenue using vectorized multiplication.

revenue = prices * quantities

print("/n Revenue for each product: ", revenue)
```

```python
# Mean and standard deviation

mean_price = np.mean(prices)

#standard deviation tells us how much the prices may vary from the average.

std_price = np.std(prices)

print("/nMean Price: ",mean_price)
print("/nStandarad Price: ", std_price)
```

```python
# Filter Products priced over $40

expensive_products = prices[prices > 40]

print("/nProducts priced above $40: ",expensive_products)
```

```python
# Reshape 12 sale values into a (4 X 3) Matrix

sales = np.array([
    10,20,30,
    40,50,60,
    70,60,90,
    100,110,120
])

print("/nOriginal Sales Array:")
print(sales)

#Convert a 1D array into a 4 X 3 Matrix.
sales_matrix = sales.reshape(4, 3)

print("/nSales Matrix (4 X 3):")
print(sales_matrix)
```
